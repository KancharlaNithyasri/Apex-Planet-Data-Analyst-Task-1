```python
import pandas as pd


path = r"C:\Users\nithish\Downloads\Task1(Apex)\Sample - Superstore.csv"
df = pd.read_csv(path, encoding='ISO-8859-1')


df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, format='mixed')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True, format='mixed')


df['Ship Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100

df.to_csv('Cleaned_one_Superstore_Data.csv', index=False)

print("Success")
print(df[['Order Date', 'Ship Date', 'Ship Days']].head(5))
```
